11. Write a program to check whether the the input number is prime or not?

```python

def is_prime(num):
    if num <= 1: return False
    for i in range(2, int(num**0.5)+1):
        if num % i == 0:
            return False
    return True

number = int(input('Enter the number'))
if is_prime(number):
    print(f'{number} is prime')
else:
    print(f'{number} is not prime')
    
```

Write the program of factorial in python ?


```python
factorial = 1
num = int(input('Enter the number'))
i = 1
while num >=i:
    factorial = factorial*i
    i+=1
print(factorial)
```
